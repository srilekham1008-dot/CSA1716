```python

from collections import deque

def water_jug(capacity1, capacity2, target):
    visited = set()
    queue = deque([(0, 0, [])])

    while queue:
        x, y, path = queue.popleft()

        if (x, y) in visited:
            continue

        visited.add((x, y))
        path = path + [(x, y)]

        if x == target or y == target:
            print("Steps to reach the target:")
            for step in path:
                print(step)
            return

        states = [
            (capacity1, y),          # Fill jug 1
            (x, capacity2),          # Fill jug 2
            (0, y),                  # Empty jug 1
            (x, 0),                  # Empty jug 2
            (x - min(x, capacity2-y),
             y + min(x, capacity2-y)),  # Pour jug 1 to jug 2
            (x + min(y, capacity1-x),
             y - min(y, capacity1-x))   # Pour jug 2 to jug 1
        ]

        for state in states:
            if state not in visited:
                queue.append((state[0], state[1], path))

    print("Target cannot be reached")

water_jug(4, 3, 2)
```

```python

from itertools import permutations

def solve_cryptarithmetic():
    letters = "SENDMORY"

    for digits in permutations(range(10), len(letters)):
        d = dict(zip(letters, digits))

        if d['S'] == 0 or d['M'] == 0:
            continue

        SEND = (d['S']*1000 + d['E']*100 +
                d['N']*10 + d['D'])

        MORE = (d['M']*1000 + d['O']*100 +
                d['R']*10 + d['E'])

        MONEY = (d['M']*10000 + d['O']*1000 +
                 d['N']*100 + d['E']*10 + d['Y'])

        if SEND + MORE == MONEY:
            print("SEND =", SEND)
            print("MORE =", MORE)
            print("MONEY =", MONEY)
            break

solve_cryptarithmetic()
```

```python

from collections import deque

def missionaries_cannibals():
    start = (3, 3, 1)
    goal = (0, 0, 0)
    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        state, path = queue.popleft()
        m, c, boat = state

        if state == goal:
            print("Solution:")
            for step in path:
                print(step)
            return

        moves = [(1, 0), (2, 0), (0, 1),
                 (0, 2), (1, 1)]

        for dm, dc in moves:
            if boat == 1:
                new_state = (m-dm, c-dc, 0)
            else:
                new_state = (m+dm, c+dc, 1)

            nm, nc, nb = new_state

            if not (0 <= nm <= 3 and 0 <= nc <= 3):
                continue

            if (nm > 0 and nm < nc) or \
               (3-nm > 0 and 3-nm < 3-nc):
                continue

            if new_state not in visited:
                visited.add(new_state)
                queue.append((new_state, path + [new_state]))

        # Continue searching until a solution is found

missionaries_cannibals()
```

```python

def vacuum_cleaner():
    rooms = {'A': 'Dirty', 'B': 'Dirty'}
    location = 'A'

    while 'Dirty' in rooms.values():
        print("Vacuum is in room", location)

        if rooms[location] == 'Dirty':
            rooms[location] = 'Clean'
            print("Room", location, "cleaned")

        elif location == 'A':
            location = 'B'
        else:
            location = 'A'

    print("All rooms are clean!")
    print(rooms)

vacuum_cleaner()
```

```python

from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        node = queue.popleft()

        if node not in visited:
            print(node, end=" ")
            visited.add(node)

            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

print("BFS Traversal:")
bfs(graph, 'A')
```

```python

def dfs(graph, node, visited):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)

        for neighbor in graph[node]:
            dfs(graph, neighbor, visited)

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

print("DFS Traversal:")
dfs(graph, 'A', set())
```

```python

from itertools import permutations

def tsp(graph, start):
    cities = list(graph.keys())
    cities.remove(start)

    min_cost = float('inf')
    best_path = []

    for path in permutations(cities):
        route = [start] + list(path) + [start]
        cost = 0

        for i in range(len(route) - 1):
            cost += graph[route[i]][route[i + 1]]

        if cost < min_cost:
            min_cost = cost
            best_path = route

    print("Shortest Path:", best_path)
    print("Minimum Cost:", min_cost)

graph = {
    'A': {'A': 0, 'B': 10, 'C': 15, 'D': 20},
    'B': {'A': 10, 'B': 0, 'C': 35, 'D': 25},
    'C': {'A': 15, 'B': 35, 'C': 0, 'D': 30},
    'D': {'A': 20, 'B': 25, 'C': 30, 'D': 0}
}

tsp(graph, 'A')
```

```python

def astar(graph, heuristic, start, goal):
    open_list = [(start, 0)]
    g_cost = {start: 0}
    parent = {}

    while open_list:
        open_list.sort(key=lambda x: x[1])
        current, _ = open_list.pop(0)

        if current == goal:
            path = []
            while current in parent:
                path.append(current)
                current = parent[current]
            path.append(start)
            path.reverse()
            print("Shortest Path:", path)
            print("Cost:", g_cost[goal])
            return

        for neighbor, cost in graph[current].items():
            new_cost = g_cost[current] + cost

            if neighbor not in g_cost or new_cost < g_cost[neighbor]:
                g_cost[neighbor] = new_cost
                priority = new_cost + heuristic[neighbor]
                parent[neighbor] = current
                open_list.append((neighbor, priority))

    print("No path found")

graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 2, 'E': 5},
    'C': {'E': 1},
    'D': {'G': 5},
    'E': {'G': 3},
    'G': {}
}

heuristic = {
    'A': 7, 'B': 6, 'C': 4,
    'D': 4, 'E': 2, 'G': 0
}

astar(graph, heuristic, 'A', 'G')
```
