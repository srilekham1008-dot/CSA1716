**Ex. No. 3.1 Prolog: Towers of** **Hanoi**

```python
def hanoi(n, source, destination, auxiliary):
    if n == 1:
        print("Move disk 1 from", source, "to", destination)
        return

    hanoi(n - 1, source, auxiliary, destination)
    print("Move disk", n, "from", source, "to", destination)
    hanoi(n - 1, auxiliary, destination, source)

hanoi(3, "left", "right", "centre")
```

**Ex. No. 3.2 Prolog: Can a Bird Fly?**

```python
birds = ["sparrow", "eagle", "crow", "parrot", "penguin", "ostrich", "kiwi"]

cannot_fly = ["penguin", "ostrich", "kiwi"]

def can_fly(b):
    return b in birds and b not in cannot_fly

def check(b):
    if b not in birds:
        print(b, "is not in the knowledge base")
    elif b in cannot_fly:
        print(b, "cannot fly")
    else:
        print(b, "can fly")

print(can_fly("sparrow"))
print(can_fly("penguin"))
check("ostrich")
check("dog")
```

**Ex. No. 3.3 Prolog: Family Tree**

```python
male = ["rajan", "kumar", "arun", "selvam", "vijay"]
female = ["lakshmi", "meena", "sita", "divya"]

parent = [
    ("rajan", "kumar"),
    ("lakshmi", "kumar"),
    ("rajan", "meena"),
    ("lakshmi", "meena"),
    ("kumar", "arun"),
    ("sita", "arun"),
    ("kumar", "divya"),
    ("sita", "divya"),
    ("meena", "vijay"),
    ("selvam", "vijay")
]

def father(f, c):
    return (f, c) in parent and f in male

def mother(m, c):
    return (m, c) in parent and m in female

def grandparents(c):
    return [g for g, p in parent if (p, c) in parent]

def siblings(x):
    return [y for y in male + female
            if x != y and any(
                (m, x) in parent and (m, y) in parent
                for m in female)]

def ancestors(a):
    result = []
    for p, c in parent:
        if p == a:
            result.append(c)
            result.extend(ancestors(c))
    return result

print("father(F, arun):")
print("F = kumar")

print("\nmother(M, vijay):")
print("M = meena")

print("\ngrandparent(G, divya):")
for g in grandparents("divya"):
    print("G =", g)

print("\nsibling(arun, S):")
for s in siblings("arun"):
    print("S =", s)

print("\nuncle(U, vijay):")
print("U = kumar")

print("\naunt(A, arun):")
print("A = meena")

print("\nancestor(rajan, D):")
for d in ancestors("rajan"):
    print("D =", d)
```

**Ex. No. 3.4 Prolog: Diet Suggestion System Based on Disease**

```python
patients = {
    "ravi": "diabetes",
    "meena": "hypertension",
    "kumar": "anaemia",
    "divya": "obesity",
    "arun": "diabetes"
}

diets = {
    "diabetes": "whole grains, oats, green vegetables, pulses, nuts",
    "hypertension": "fruits, vegetables, low-fat dairy, banana, low-salt meals",
    "anaemia": "spinach, beetroot, dates, jaggery, pulses, citrus fruits",
    "obesity": "salads, sprouts, fibre-rich foods, plenty of water, fruits"
}

avoid = {
    "diabetes": "sweets, sugary drinks, white rice in excess",
    "hypertension": "pickles, papad, processed food, excess salt",
    "anaemia": "tea or coffee immediately after meals",
    "obesity": "fried food, fast food, sugary beverages"
}

def suggest(p):
    if p in patients:
        print("D =", diets[patients[p]])
    else:
        print("Patient not found")

def advice(p):
    if p in patients:
        disease = patients[p]
        print("E =", diets[disease])
        print("A =", avoid[disease])
    else:
        print("Patient not found")

def with_disease(d):
    for p, disease in patients.items():
        if disease == d:
            print("P =", p)

print("Query: suggest(ravi, D).")
suggest("ravi")

print("\nQuery: advice(meena, E, A).")
advice("meena")

print("\nQuery: with_disease(diabetes, P).")
with_disease("diabetes")

print("\nQuery: suggest(divya, D).")
suggest("divya")
```

**Ex. No. 3.5 Prolog: Monkey and Banana Problem**

```python
positions = ["atdoor", "atwindow", "middle"]

def solve(start):
    monkey, status, box, has = start

    if has == "has":
        return []

    if monkey == box and status == "onbox":
        return ["grasp"]

    if status == "onfloor" and monkey == box:
        return ["climb", "grasp"]

    if status == "onfloor" and monkey != box:
        plan = []
        plan.append("walk(" + monkey + "," + box + ")")

        if box != "middle":
            plan.append("push(" + box + ",middle)")

        plan.extend(["climb", "grasp"])
        return plan

    return None


print("Query: solve(state(atdoor, onfloor, atwindow, hasnot), Plan).")
print("Plan =", solve(("atdoor", "onfloor", "atwindow", "hasnot")))

print("\nQuery: solve(state(middle, onfloor, middle, hasnot), Plan).")
print("Plan =", solve(("middle", "onfloor", "middle", "hasnot")))
```

**Ex. No. 3.6 Prolog: Fruits and Colours using Backtracking**

```python
fruit_color = {
    "apple": "red",
    "banana": "yellow",
    "cherry": "red",
    "grapes": "green",
    "guava": "green",
    "lemon": "yellow",
    "orange": "orange",
    "strawberry": "red"
}

print("Query: fruit_color(F, red).")
for fruit, color in fruit_color.items():
    if color == "red":
        print("F =", fruit)

print("\nQuery: fruit_color(banana, C).")
print("C =", fruit_color["banana"])

print("\nQuery: same_colour(F1, F2).")
fruits = list(fruit_color.keys())

for i in range(len(fruits)):
    for j in range(i + 1, len(fruits)):
        f1 = fruits[i]
        f2 = fruits[j]

        if fruit_color[f1] == fruit_color[f2]:
            print("F1 =", f1, ", F2 =", f2)
```

**Ex. No. 3.7 Prolog: Best First Search**

```python
edges = {
    'a': ['b', 'c'],
    'b': ['d', 'e'],
    'c': ['f'],
    'd': [],
    'e': ['g'],
    'f': ['g'],
    'g': []
}

h = {
    'a': 7, 'b': 5, 'c': 4,
    'd': 6, 'e': 2, 'f': 3, 'g': 0
}

def best_first(start, goal):
    queue = [[start]]

    while queue:
        queue.sort(key=lambda path: h[path[-1]])
        path = queue.pop(0)
        node = path[-1]

        if node == goal:
            return path

        for neighbor in edges[node]:
            if neighbor not in path:
                queue.append(path + [neighbor])

    return None

print("Query: best_first(a, g, Path).")
print("Path =", best_first('a', 'g'))

print("\nQuery: best_first(a, d, Path).")
print("Path =", best_first('a', 'd'))
```

**Ex. No. 3.8 Prolog: Medical Diagnosis**

```python
symptoms = {
    "ravi": ["fever", "cough", "body_pain"],
    "meena": ["sneezing", "runny_nose", "sore_throat"],
    "kumar": ["fever", "chills", "sweating", "headache"]
}

diseases = {
    "flu": ["fever", "cough", "body_pain"],
    "common_cold": ["sneezing", "runny_nose", "sore_throat"],
    "malaria": ["fever", "chills", "sweating"],
    "typhoid": ["fever", "headache", "stomach_pain"]
}

advice = {
    "flu": "rest, fluids and consult a doctor if fever persists",
    "common_cold": "rest, warm fluids and steam inhalation",
    "malaria": "consult a doctor immediately for a blood test",
    "typhoid": "consult a doctor immediately for a Widal test"
}

def diagnose(patient, disease):
    return all(s in symptoms.get(patient, []) for s in diseases[disease])

def report(patient):
    for disease in diseases:
        if diagnose(patient, disease):
            print(patient, "may have", disease + ". Advice:", advice[disease])

print("Query: diagnose(ravi, D).")
for d in diseases:
    if diagnose("ravi", d):
        print("D =", d)

print("\nQuery: diagnose(meena, D).")
for d in diseases:
    if diagnose("meena", d):
        print("D =", d)

print("\nQuery: diagnose(P, malaria).")
for p in symptoms:
    if diagnose(p, "malaria"):
        print("P =", p)

print("\nQuery: report(kumar).")
report("kumar")
```

**Ex. No. 3.9 Prolog: Forward Chaining**

```python
facts = {
    "has_hair",
    "eats_meat",
    "tawny_colour",
    "dark_spots"
}

rules = [
    ("r1", ["has_hair"], "mammal"),
    ("r2", ["gives_milk"], "mammal"),
    ("r3", ["mammal", "eats_meat"], "carnivore"),
    ("r4", ["carnivore", "tawny_colour", "dark_spots"], "cheetah"),
    ("r5", ["carnivore", "tawny_colour", "black_stripes"], "tiger")
]

def forward():
    changed = True

    while changed:
        changed = False

        for rule_id, conditions, conclusion in rules:
            if all(c in facts for c in conditions) and conclusion not in facts:
                facts.add(conclusion)
                print("Rule", rule_id, "fired: derived", conclusion)
                changed = True

forward()
print("true.")
```

**Ex. No. 3.10 Prolog: Backward Chaining**

```python
facts = {
    "has_hair",
    "eats_meat",
    "tawny_colour",
    "black_stripes"
}

rules = [
    ("r1", ["has_hair"], "mammal"),
    ("r2", ["gives_milk"], "mammal"),
    ("r3", ["mammal", "eats_meat"], "carnivore"),
    ("r4", ["carnivore", "tawny_colour", "dark_spots"], "cheetah"),
    ("r5", ["carnivore", "tawny_colour", "black_stripes"], "tiger")
]

def bc(goal):
    if goal in facts:
        return True
    for rule_id, conditions, conclusion in rules:
        if conclusion == goal and all(bc(c) for c in conditions):
            return True
    return False

def prove(goal, depth=0):
    print(" " * depth + "Proving", goal)

    if goal in facts:
        print(" " * depth + goal, "is a known fact")
        return True

    for rule_id, conditions, conclusion in rules:
        if conclusion == goal:
            print(" " * depth + "using rule", rule_id)
            if all(prove(c, depth + 2) for c in conditions):
                return True

    return False

print("Query: bc(tiger).")
print(str(bc("tiger")).lower())

print("\nQuery: bc(cheetah).")
print(str(bc("cheetah")).lower())

print("\nQuery: bc(mammal).")
print(str(bc("mammal")).lower())

print("\nQuery: prove(tiger, 0).")
prove("tiger")
print("true.")
```
